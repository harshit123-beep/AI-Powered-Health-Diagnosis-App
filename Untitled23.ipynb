```python
import pandas as pd
import numpy as np
import speech_recognition as sr
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.decomposition import TruncatedSVD
```

```python
movies = pd.read_csv('movies.csv')
movies["combined"] = (
    movies["genres"].fillna("") + " " +
    movies["keywords"].fillna("")
).str.lower()
print(movies["combined"].head())
```

```python
tfidf = TfidfVectorizer(stop_words="english")
tfidf_matrix = tfidf.fit_transform(movies["combined"])

content_similarity = cosine_similarity(tfidf_matrix)
print(content_similarity)
```

```python
user_item_matrix = ratings.pivot(index="userId", columns="movieId", values="rating").fillna(0)

svd = TruncatedSVD(n_components=2, random_state=42)
reduced = svd.fit_transform(user_item_matrix)
matrix_reconstructed = np.dot(reduced, svd.components_)
user_item_matrix.head()
```

```python
def content_recommend(title):
    if title not in movies["title"].values:
        return ["Movie not found"]

    idx = movies[movies["title"] == title].index[0]
    scores = list(enumerate(content_similarity[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)[1:6]

    return [movies.iloc[i[0]]["title"] for i in scores]
recs = content_recommend("Inception")

for i, movie in enumerate(recs, 1):
    print(f"{i}. {movie}")
```

```python
def hybrid_recommend(user_id, title):
    content_list = content_recommend(title)

    user_idx = user_id - 1
    scores = matrix_reconstructed[user_idx]
    movie_indices = np.argsort(scores)[::-1]

    collab_list = [movies.iloc[i]["title"] for i in movie_indices[:5]]

    final = list(set(content_list + collab_list))
    return final[:5]
    recs = hybrid_recommend(1, "Inception")

print("Recommended Movies:\n")
for i, movie in enumerate(recs, 1):
    print(f"{i}. {movie}")
```

```python
from IPython.display import Javascript
from google.colab import output
import speech_recognition as sr
import io
from pydub import AudioSegment

def record_audio():
    js = Javascript('''
    async function record() {
        const div = document.createElement('div');
        const button = document.createElement('button');
        button.textContent = 'Start Recording';
        div.appendChild(button);

        document.body.appendChild(div);

        const stream = await navigator.mediaDevices.getUserMedia({audio: true});
        const recorder = new MediaRecorder(stream);
        let chunks = [];

        recorder.ondataavailable = e => chunks.push(e.data);

        button.onclick = () => {
            recorder.start();
            setTimeout(() => recorder.stop(), 3000);
        };

        recorder.onstop = async () => {
            const blob = new Blob(chunks);
            const arrayBuffer = await blob.arrayBuffer();
            const base64 = btoa(
                String.fromCharCode(...new Uint8Array(arrayBuffer))
            );
            google.colab.kernel.invokeFunction('notebook.recorded_audio', [base64], {});
        };
    }
    record();
    ''')
    display(js)

audio_data = None

def save_audio(base64):
    global audio_data
    audio_data = base64

output.register_callback('notebook.recorded_audio', save_audio)
```

```python
import base64
import speech_recognition as sr # Added import for sr

def speech_to_text():
    global audio_data
    if audio_data is None:
        return "No audio recorded"

    audio_bytes = base64.b64decode(audio_data)

    with open("audio.wav", "wb") as f:
        f.write(audio_bytes)

    r = sr.Recognizer()
    with sr.AudioFile("audio.wav") as source:
        audio = r.record(source)

    try:
        text = r.recognize_google(audio)
        print("You said:", text)
        return text
    except:
        return "Inception"

movie_name = speech_to_text() # Moved this line outside the function
print(movie_name)
```

```python
def detect_emotion(text):
    text = text.lower()

    if "sad" in text:
        return "sad"
    elif "happy" in text:
        return "happy"
    elif "love" in text:
        return "romantic"
    elif "excited" in text:
        return "excited"
    else:
        return "neutral"
```

```python
user_id = 1

voice_text = speech_to_text()

emotion = detect_emotion(voice_text)

print("Detected Emotion:", emotion)

print("Recommendations:")
print(hybrid_recommend(user_id, "Inception"))
```

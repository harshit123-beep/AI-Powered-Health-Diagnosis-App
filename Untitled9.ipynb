```python
import os
from ultralytics import YOLO

# Load pretrained YOLOv8 model
model = YOLO("yolov8n.pt")  # n = nano (fastest)

# Run detection on an image
results = model("zidane.jpg") # Changed from car_parking.jpg to zidane.jpg

# Show results
for r in results:
    r.show()

# Save results
output_dir = "output/"
os.makedirs(output_dir, exist_ok=True)
for i, r in enumerate(results):
    save_path = os.path.join(output_dir, f"result_{i}.jpg")
    r.save(save_path)
```
